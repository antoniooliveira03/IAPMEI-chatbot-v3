```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/evaluation/Chatbot Performance Q&A.csv')
```

```python
df.columns
```

```python
df.columns = [
    "timestamp", "username",
    "Q1", "Q2", "Q3",
    "Q4", "Q5", "Q6",
    "Q7", "Q8", "Q9",
    "Q10", "Q11",
    "Q12", "Q13"
]
```

```python
df.describe()
```

```python
import numpy as np
import matplotlib.pyplot as plt

mean_scores = df.iloc[:, 2:13].mean()
median_scores = df.iloc[:, 2:13].median()

x = np.arange(len(mean_scores))
width = 0.35

fig, ax = plt.subplots()

# Greens for consistency
mean_color = plt.cm.Greens(0.6)
median_color = plt.cm.Greens(0.85)

ax.bar(x - width/2, mean_scores, width, label='Mean', color=mean_color)
ax.bar(x + width/2, median_scores, width, label='Median', color=median_color)

ax.set_xticks(x)
ax.set_xticklabels(mean_scores.index, rotation=45, ha='right')
ax.set_ylabel("Score")
ax.legend(title="Score", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
likert = df.iloc[:, 2:13]
likert_percent = {}

for col in likert.columns:
    counts = likert[col].value_counts(normalize=True).sort_index() * 100
    likert_percent[col] = counts

likert_df = pd.DataFrame(likert_percent).T.fillna(0)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define order of responses
categories = [1, 2, 3, 4, 5]

# Create green color gradient
cmap = plt.cm.Greens
colors = cmap(np.linspace(0.3, 0.9, len(categories)))  

fig, ax = plt.subplots()

left = [0] * len(likert_df)

for i, cat in enumerate(categories):
    values = likert_df[cat] if cat in likert_df.columns else [0]*len(likert_df)
    #print(values)
    ax.barh(
        likert_df.index,
        values,
        left=left,
        label=f"{cat}",
        color=colors[i]
    )
    left = [l + v for l, v in zip(left, values)]

ax.set_xlabel("Percentage (%)")
ax.set_title("Response Distribution")

ax.legend(title="Score", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
