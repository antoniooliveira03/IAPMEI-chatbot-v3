```python
mean_scores = df.iloc[:, 2:13].mean()

plt.figure()
mean_scores.plot(kind='bar', color=plt.cm.Greens(0.6))

plt.xticks(rotation=0, ha='right')
plt.ylabel("Average Score")
#plt.title("Average Survey Scores per Question")

plt.tight_layout()
plt.show()
```

```python
df.iloc[:, 2:13].hist()
plt.tight_layout()
plt.show()
```

```python
corr = df.iloc[:, 2:13].corr()

plt.figure()
plt.imshow(corr)
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=90)
plt.yticks(range(len(corr)), corr.columns)
plt.title("Correlation Matrix")
plt.show()
```

```python
likert = df.iloc[:, 2:13]
likert_percent = {}

for col in likert.columns:
    counts = likert[col].value_counts(normalize=True).sort_index() * 100
    likert_percent[col] = counts

likert_df = pd.DataFrame(likert_percent).T.fillna(0)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define order of responses
categories = [1, 2, 3, 4, 5]

# Create green color gradient
cmap = plt.cm.Greens
colors = cmap(np.linspace(0.3, 0.9, len(categories)))  

fig, ax = plt.subplots()

left = [0] * len(likert_df)

for i, cat in enumerate(categories):
    values = likert_df[cat] if cat in likert_df.columns else [0]*len(likert_df)
    #print(values)
    ax.barh(
        likert_df.index,
        values,
        left=left,
        label=f"{cat}",
        color=colors[i]
    )
    left = [l + v for l, v in zip(left, values)]

ax.set_xlabel("Percentage (%)")
ax.set_title("Response Distribution")

ax.legend(title="Score", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
